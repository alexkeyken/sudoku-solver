```python
BOARD_SIZE = 9 # the board contains BOARD_SIZE * BOARD_SIZE digits
BOX_SIZE = 3 # each box contains BOX_SIZE * BOX_SIZE digits
BOXES_PER_ROW = BOARD_SIZE//BOX_SIZE

def get_box_digits(board: list[list[int]], cell: tuple[int, int]) -> list[int]:
    """
    Retrieves a list containing all digits inside the box that the cell belongs to. 
    """
    box_digit_list = []
    row, col = cell
    box_row_offset, box_col_offset = row//BOX_SIZE, col//BOX_SIZE
    for r in range(BOX_SIZE):
        for k in range(BOX_SIZE):
            box_digit_list.append(board[BOX_SIZE*box_row_offset+r][BOX_SIZE*box_col_offset+k])
    
    return box_digit_list

def compute_possible_cell_digits(board: list[list[int]], cell: tuple[int, int]) -> list[int]:
    """
    Computes all digits that would not cause rule breaks in this cell.
    """
    possible_digit_list = []
    row, col = cell
    for digit in range(1, BOX_SIZE*BOX_SIZE+1):
        is_in_row = digit in board[row]
        is_in_col = digit in [board[row_index][col] for row_index in range(len(board))]
        is_in_box = digit in get_box_digits(board, cell)

        if not (is_in_row or is_in_col or is_in_box):
            possible_digit_list.append(digit)

    return possible_digit_list

def compute_possible_board_digits(board: list[list[int]]) -> dict[tuple[int, int] : list[int]]:
    """
    Creates a dictionary containing lists of all possible digits for each cell,
    indexed by the cell coordinates: dict[(row,col)] = [list of digits].
    """
    possible_digit_dict = {}
    for row in range(len(board)):
        for col in range(len(board[row])):
            if board[row][col] == 0:
                cell = (row, col)
                possible_digit_dict[cell] = compute_possible_cell_digits(board, cell)
                
    return possible_digit_dict

def update_board(board: list[list[int]], possible_digit_dict: dict[tuple[int, int] : list[int]]) -> None:
    """
    Updates empty cells in board where only one digit would be possible.
    """
    for cell, possible_digits in possible_digit_dict.items():
        if len(possible_digits) == 1:
            row, col = cell
            board[row][col] = possible_digits[0]

def check_solved_board(board: list[list[int]]) -> bool:
    """
    Check completed board for repeated digits within each row, column and subgrid.
    """
    for row in board:
        if len(row) != len(set(row)):
            print(f"Repeated digit in row {row}.")
            return False

    transposed_board = list(map(list, zip(*board)))
    for col in transposed_board:
        if len(col) != len(set(col)):
            print(f"Repeated digit in column {col}.")
            return False

    for subgrid_row in range(BOXES_PER_ROW):
        for subgrid_col in range(BOXES_PER_ROW):
            subgrid = get_box_digits(board, (BOX_SIZE*subgrid_row, BOX_SIZE*subgrid_col))
            if len(subgrid) != len(set(subgrid)):
                print(f"Repeated digit in subgrid [{subgrid_row}, {subgrid_col}].")
                return False     
                
    return True
```
